```python
path = "/content/train (2).csv"
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer
from sklearn.metrics import accuracy_score
```

```python

df = pd.read_csv(path)
df=df[['Age','Pclass','Fare','Survived']]
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
x = df.drop("Survived",axis=1)
```

```python
y = df[["Survived"]]
```

```python
x_train ,x_test ,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = LogisticRegression()
```

```python
imputer = SimpleImputer(strategy="mean")
```

```python
imputer_knn = KNNImputer(n_neighbors= 5,weights="distance")
```

```python
x_train_simple =  imputer.fit_transform(x_train)
x_test_simple = imputer.fit_transform(x_test)
x_train_knn= imputer_knn.fit_transform(x_train)
x_test_knn = imputer_knn.fit_transform(x_test)
```

```python
model.fit(x_train_simple,y_train)
```

```python
pred = model.predict(x_test_simple)
```

```python
print("simpleImputer:",accuracy_score(pred ,y_test))
```

```python
model.fit(x_train_knn,y_train)
```

```python
pred2 = model.predict(x_test_knn)
```

```python
print("KNN:",accuracy_score(y_test,pred2))
```

Compared SimpleImputer and KNNImputer for handling missing values. KNNImputer achieved a higher score (0.7542 vs 0.7374), indicating that using neighboring samples provided better estimates for missing data. Therefore, KNNImputer was selected for further modeling.
