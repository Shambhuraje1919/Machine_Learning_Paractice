```python
import pandas as pd
import seaborn as sb
import numpy as np
import matplotlib.pyplot as plt

```

```python


brand_counts = dp["Brand"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    brand_counts,
    labels=brand_counts.index,
    autopct="%1.1f%%",
    shadow=True,
    startangle=90
)

plt.title("Brand Distribution")

plt.show()
```
