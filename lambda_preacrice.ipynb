```python
students = [
    ("Sam", 85),
    ("Raj", 92),
    ("Amit", 78)
]

students.sort(key=lambda x: x[1])

print(students)
```

```python
students = [
    ("Sam", 85),
    ("Raj", 92),
    ("Amit", 78)
]

topper = max(students, key=lambda x: x[1])

print(topper)
```

```python
check = lambda age: "Adult" if age >= 18 else "Minor"

print(check(20))
print(check(15))
```

```python
check = lambda price: "high" if price >=10000 else "low"
print(check(9000))
print(check(110000))
```

```python
words = ["apple", "kiwi", "banana", "mango"]

words.sort(key=lambda x: len(x))

print(words)
```

```python
nums = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * 2, nums))

print(result)
```
