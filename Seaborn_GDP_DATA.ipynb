```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
data_root = "http://github.com/ageron/data/raw/main/"
```

```python
life = pd.read_csv(data_root + "lifesat/lifesat.csv")
```

```python
life.head(2)
```

```python
x = life[['GDP per capita (USD)']].values
```

```python
y = life[['Life satisfaction']].values
```

```python
life.plot(kind='scatter',grid=True,x ="GDP per capita (USD)",y='Life satisfaction')
plt.axis([23_500,62_500,4,9])
plt.show()
```

```python
model = LinearRegression()
```

```python
model.fit(x,y)
```

```python
x_new = [[33_442.8]] # new desh chi gdp
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(model.predict(x_new))
```
