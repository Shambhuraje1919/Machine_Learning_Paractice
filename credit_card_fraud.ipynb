```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("uditjain13/credit-card-fraud-detection-2026")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as pl
```

```python
os.listdir(path)
```

```python
df = pd.read_csv(os.path.join(path,'credit_card_fraud_2026.csv'))
```

```python
fig = pl.scatter_3d(df, x='time_of_day_hour',y = 'hours_since_last_txn',z =  'cvv_retry_count',color='is_fraud')
fig.show()
```

```python
pl.line_3d(df, x='ip_country_mismatch',y = 'hours_since_last_txn',z =  'cvv_retry_count',color='is_fraud')
```

```python
df.dtypes
```

```python
df= pd.get_dummies(df, columns= ['is_ai_generated_scam_attempt','billing_shipping_mismatch','ip_country_mismatch','used_vpn','is_new_merchant','is_foreign_transaction'],drop_first=True)
```

```python
df = pd.get_dummies(df, columns=['card_type','auth_method','channel','device_type',],drop_first=True)
```

```python
df = pd.get_dummies(df,columns=['merchant_category'],drop_first=True)
```

```python
df.dtypes
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score , confusion_matrix
from sklearn.ensemble import RandomForestClassifier
```

```python
x = df.drop('is_fraud',axis=1)
```

```python
y = df['is_fraud']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,random_state=42, test_size=0.2)
```

```python
model = LogisticRegression()
model2 = RandomForestClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
model2.fit(x_train,y_train)
```

```python
df.shape
```

```python
pred = model.predict(x_test)
pred1 = model2.predict(x_test)
```

```python
pred1
```

```python
pred
```

```python
print('accuracy_score Logistic', accuracy_score(y_test,pred))
```

```python
print('accuracy_score RandomForest', accuracy_score(y_test,pred1))
```

```python
from xgboost import XGBClassifier

model3 = XGBClassifier(random_state=42)
model3.fit(x, y)

importance = pd.DataFrame({
    "Feature": x.columns,
    "Importance": model3.feature_importances_
})

importance = importance.sort_values(by="Importance", ascending=False)
print(importance.head(20))
```

```python
model2.fit(x, y)

importance = pd.DataFrame({
    "Feature": x.columns,
    "Importance": model2.feature_importances_
})

importance = importance.sort_values(by="Importance", ascending=False)
print(importance.head(20))
```
