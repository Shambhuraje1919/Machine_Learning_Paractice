```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, confusion_matrix, mean_absolute_percentage_error
from sklearn.ensemble import RandomForestRegressor
```

```python
si = SimpleImputer(strategy='mean')
```

```python
x = df.drop('Amount',axis =1)
```

```python
y = df['Amount']
```

```python
x_train , x_test , y_train ,y_test = train_test_split(x,y,train_size=0.2,random_state=44)
```

```python
x_train = si.fit_transform(x_train)
```

```python
x_test = si.transform(x_test)
```

```python
model = LinearRegression()
```

```python
model2 = RandomForestRegressor()
```

```python
model.fit(x_train,y_train)
```

```python
Y_pred = model.predict(x_test)
```

```python
print(mean_absolute_error(Y_pred,y_test))
```

```python
print(mean_absolute_percentage_error(Y_pred,y_test))
```
