```python
from torchvision import models
from torchvision import transforms
```

```python
dir(models)
```

```python
alexnet = models.AlexNet()
```

```python
from torchvision.models import ResNet101_Weights
resnet = models.resnet101(weights=ResNet101_Weights.DEFAULT)
```

```python
resnet
```

```python
preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean = [0.485,0.456 , 0.406],
        std = [0.229,0.224,0.225]
    )
])
```

```python
from PIL import Image
```

```python
img = Image.open('/content/goldfish.jpg')
```

```python
img_t = preprocess(img)
```

```python
import torch
```

```python
batch_t = torch.unsqueeze(img_t,0)
```

```python
resnet.eval()
```

```python
out = resnet(batch_t)
```

```python
out
```

```python
from torchvision.models import ResNet50_Weights
weights = ResNet50_Weights.DEFAULT
labels = weights.meta["categories"]


print(labels[:5])
len(labels)
```

```python
_, index = torch.max(out,1)
```

```python
precen = torch.nn.functional.softmax(out,dim = 1)[0]*100
labels[index[0]], precen[index[0]].item()
```

```python
top5_prob, top5_indices = torch.topk(precen, 5)

print("--- Top 5 Predictions ---")
for i in range(5):
    idx = top5_indices[i].item()
    label = labels[idx]
    confidence = top5_prob[i].item()
    print(f"{label}: {confidence:.2f}%")
```
