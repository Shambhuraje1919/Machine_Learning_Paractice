```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("yasserh/titanic-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
os.listdir(path)
```

```python
original = pd.read_csv(os.path.join(path,'Titanic-Dataset.csv'))
```

```python
original.head()
```

```python
df = original.drop(columns=['PassengerId','Name','Ticket','Fare'],axis=1)
```

```python
df.head()
```

```python
df['family'] = df['SibSp']+df['Parch']
```

```python
df.head()
```

```python
df =df.drop(columns=['SibSp','Parch'],axis=1)
```

```python
df = df.drop('Cabin',axis=1)
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
## missing value fill up
upper = df['Age'].mean()+(3*df['Age'].std())
lower = df['Age'].mean()-(3*df['Age'].std())
si = SimpleImputer(strategy='mean')
sii = SimpleImputer(strategy='most_frequent')
print(upper," ",lower)
```

```python
df.isnull().mean()*100
```

```python
df['Age'] = si.fit_transform(df[['Age']])
```

```python
df[['Embarked']] =sii.fit_transform(df[['Embarked']])
```

```python
imputer = SimpleImputer(strategy="mean")
le = LabelEncoder()
ohe = OneHotEncoder(sparse_output=False)
```

```python
df['Sex'] = le.fit_transform(df['Sex'])
```

```python
data = ohe.fit_transform(df[['Embarked']])
new_cat = ohe.get_feature_names_out(['Embarked'])
df[new_cat]=data
```

```python
df = df.drop('Embarked',axis=1)
```

```python
df.isnull().mean()*100
```

```python
x = df.drop('Survived',axis=1)
```

```python
x.head()
```

```python
y = df[['Survived']]
```

```python
y.head()
```

```python
x_train ,x_test ,y_train ,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model1= LogisticRegression()
model2 =RandomForestClassifier(n_estimators=10,max_depth=50)
```

```python
model1.fit(x_train,y_train)
model2.fit(x_train,y_train)
```

```python
pred1 = model1.predict(x_test)
pred2 = model2.predict(x_test)
```

```python
print(accuracy_score(y_test,pred1),"Logistic")
print(accuracy_score(y_test,pred2),"Random_forest")
```

```python
print(confusion_matrix(y_test,pred1))
```

```python
print(confusion_matrix(y_test,pred2))
```
