```python
import plotly as px
import pandas as pd
```

```python
data = {
    "marks":[10,20,30,40,50],
    "passing":[36,36,36,36,36]
}
```

```python
px.plot(data,kind= "bar")
```

```python
df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [100, 120, 90, 150, 180]
})

fig = px.plot(df, x="Day", y="Sales", title="Sales Trend",kind='line')
fig.show()
```

```python
df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [30000, 450000, 50000, 70000, 85000]
})

fig = px.plot(
    df,
    x="Age",
    y="Salary",
    title="Age vs Salary",
    kind='scatter'
)

fig.show()
```

```python
px.plot(df,x='Age',y='Salary',kind='hist')
```
