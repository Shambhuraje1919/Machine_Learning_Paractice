```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
from sklearn.metrics import confusion_matrix, confusion_matrix , classification_report
```

```python
x = df.drop("placement_status_Placed",axis=1)
x_new = df.drop(columns=["placement_status_Placed","exam_score"],axis=1)
```

```python
df = pd.get_dummies(df,"placement_status",drop_first=True)
```

```python
df.columns
```

```python
y = df["placement_status_Placed"]
```

```python
x_train,x_test,y_train , y_test = train_test_split(x_new,y,test_size = 0.2,random_state=42)
```

```python
models = {
"model1" : LogisticRegression(),
"model2" : DecisionTreeClassifier(),
"model3" : RandomForestClassifier(),
"model4" : KNeighborsClassifier(),
"model5" : XGBClassifier(),
}
```

```python
for name ,model in models.items():
  model.fit(x_train,y_train)
  train_score =model.score(x_train ,y_train)
  test_score = model.score(x_test,y_test)
  print(model)
  print("Train Accuracy:",train_score)
  print("Test Accuracy :", test_score)
  print("total diff",(test_score-train_score))
  y_pred = model.predict(x_test)
  print(confusion_matrix(y_pred,y_test))

```
