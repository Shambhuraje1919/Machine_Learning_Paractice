```python
import polars as pl
```

```python
data = {
    "restaurant_name": ["Tandoor House", "Copper Deck", "Biryani By Kilo", "Spice Junction", "Urban Cafe"],
    "city": ["Chennai", "Delhi", "Delhi", "Pune", "Delhi"],
    "swiggy_rating": [4.0, 3.9, 4.4, 4.1, 3.6],
    "zomato_rating": [3.7, 4.1, 4.5, 4.0, 3.8],
    "avg_cost": [310, 450, 600, 250, 200]
}
```

```python
df = pl.DataFrame(data)
```

```python
result = df.filter((pl.col('city')=='Delhi') & (pl.col('avg_cost') > 300)).select(['restaurant_name','zomato_rating','swiggy_rating'])
```

```python
result
```

```python
result = df.group_by("city").agg([
    pl.col("zomato_rating").mean().alias("avg_zomato"),
    pl.len().alias("total_count")
])

print(result)
```

```python
res = df.with_columns(((pl.col('swiggy_rating')+ pl.col('zomato_rating'))/2).alias('Overall_rating')).filter(pl.col('Overall_rating')>4)
```

```python
res
```

```python
result = df.with_columns(
    pl.when(pl.col("avg_cost") > 400)
    .then(pl.lit("High"))
    .when(pl.col("avg_cost") >= 250)
    .then(pl.lit("Medium"))
    .otherwise(pl.lit("Low"))
    .alias("cost_category")
)

print(result)
```
