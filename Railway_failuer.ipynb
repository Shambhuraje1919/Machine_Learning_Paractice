```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
```

```python
model1 = LogisticRegression()
```

```python
x = df.drop('maintenance_required',axis=1)
```

```python
y = df[['maintenance_required']]
```

```python
x_train , x_test , y_train,y_test = train_test_split(x,y,test_size=0.2,random_state= 44)
```

```python
model1.fit(x_train,y_train)
```

```python
y_pred = model1.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("accuracy" ,accuracy_score(y_test,y_pred))
```

```python
model2 = RandomForestClassifier(n_estimators=45,max_depth=7)
```

```python
from sklearn.decomposition import PCA

print(type(x_train))
print(type(x_test))

print(x_train.shape)
print(x_test.shape)

pca = PCA(n_components=3)

x_train_pca = pca.fit_transform(x_train)

print("n_features_in_:", pca.n_features_in_)
print("components shape:", pca.components_.shape)
```

```python
x_train_pca = pca.fit_transform(x_train)
x_test_pca = pca.transform(x_test)

print(x_train_pca.shape)
print(x_test_pca.shape)
```

```python
import pandas as pd

pca_df = pd.DataFrame(
    x_train_pca,
    columns=["PC1", "PC2", "PC3"]
)

pca_df["maintenance_required"] = y_train.reset_index(drop=True)
```

```python
import plotly.express as px

fig = px.scatter_3d(
    pca_df,
    x="PC1",
    y="PC2",
    z="PC3",
    color="maintenance_required",
    color_discrete_map={
        0: "green",
        1: "red"
    },
    opacity=0.7,
    title="3D PCA Visualization - Railway Maintenance Prediction",
    labels={"maintenance_required": "Maintenance"}
)

fig.update_traces(marker=dict(size=3))

fig.update_layout(
    template="simple_white",
    scene=dict(
        xaxis_title="Principal Component 1",
        yaxis_title="Principal Component 2",
        zaxis_title="Principal Component 3"
    ),
    width=900,
    height=700
)

fig.show()
```
