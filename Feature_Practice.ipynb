```python
import pandas as pd
from io import StringIO

data = """
Age,Gender,City,Salary,Experience,Married,Purchased
22,Male,Pune,25000,1,No,0
35,Female,Mumbai,50000,8,Yes,1
28,Male,Nashik,32000,3,No,0
45,Female,Pune,80000,15,Yes,1
31,Male,,45000,5,Yes,1
26,Female,Mumbai,,2,No,0
38,Male,Nashik,60000,,Yes,1
29,Female,Pune,40000,4,,0
50,Male,Mumbai,90000,20,Yes,1
24,Female,Nashik,28000,1,No,0
42,Male,Pune,75000,12,Yes,1
33,Female,,52000,7,No,1
27,Male,Mumbai,35000,3,No,0
48,Female,Nashik,85000,18,Yes,1
30,Male,Pune,,5,No,0
36,Female,Mumbai,58000,9,Yes,1
25,Male,Nashik,30000,,No,0
41,Female,Pune,72000,11,Yes,1
32,Male,Mumbai,47000,6,,1
29,Female,Nashik,39000,4,No,0
"""

df = pd.read_csv(StringIO(data))

df.head()
```

```python
df.isnull()
```

```python
df.fillna(df.mean(numeric_only=True))
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
LE = LabelEncoder()
```

```python
df["Gender"] = LE.fit_transform(df["Gender"])
```

```python
df["Married"] = LE.fit_transform(df["Married"])
```

```python
df["Married"].head(3)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder()
```

```python
encoded = encoder.fit_transform(df[["City"]])
```

```python
encoded_df = pd.DataFrame(encoded.toarray(),columns=encoder.get_feature_names_out())
```

```python
df = pd.concat([df,encoded_df],axis=1)
```

```python
df.drop("City",axis=1,inplace=True)
```

```python
df = df.loc[:, ~df.columns.duplicated()]
```

```python
df["Experience"] = df["Experience"].fillna(df["Experience"].mean())
```

```python
df.sample()
```

```python
df["salary_per_exp"] = (df["Salary"]/ df["Experience"]+1)
```

```python
df.sample()
```

```python
x = df.drop("Purchased",axis=1)
```

```python
y = df["Purchased"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train ,y_test = train_test_split(x,y , train_size= 0.2,random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
pred =model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(pred, y_test))
```
