```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
scaler = StandardScaler()
```

```python
x = df.drop('quality',axis=1)
```

```python
y = df['quality']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size = 0.2, random_state=23)
```

```python
x_train.head(3)
```

```python
y_train.sample(2)
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
y_train[:2]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_absolute_percentage_error
from sklearn.ensemble import RandomForestRegressor
```

```python
pipeline = ([('scaler', StandardScaler()),('linear', LinearRegression())])
```

```python
models = {'linea' : LinearRegression(),
"Random_forest" : RandomForestRegressor(n_estimators=200,
    max_depth=12,
    min_samples_split=5,
    random_state=42)}
```

```python
for name ,model in models.items():
  model.fit(x_train , y_train)
  y_pred = model.predict(x_test)
  print(name,"r2_score:-",r2_score(y_test,y_pred))
  print(name, "mean_absolute_percentage_error:-",mean_absolute_percentage_error(y_test,y_pred))
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=3)
```

```python
x_train_pca = pca.fit_transform(x_train)
x_test_pca = pca.transform(x_test)
```

```python
import plotly.express as px

# Create a clean DataFrame from your 3D PCA array
plot_df = pd.DataFrame({
    'PC1': x_train_pca[:, 0],
    'PC2': x_train_pca[:, 1],
    'PC3': x_train_pca[:, 2],
    'Quality': y_train
})

# Generate interactive 3D scatter plot
fig = px.scatter_3d(
    plot_df,
    x='PC1',
    y='PC2',
    z='PC3',
    color='Quality',
    color_continuous_scale='Plasma',
    title='Interactive 3D PCA - Wine Quality'
)

# Update layout for better sizing
fig.update_layout(
    margin=dict(l=0, r=0, b=0, t=40),
    scene=dict(
        xaxis_title='PC 1',
        yaxis_title='PC 2',
        zaxis_title='PC 3'
    )
)

fig.show()
```
