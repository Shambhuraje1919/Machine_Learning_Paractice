```python
import pandas as  pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {
    "x":[1,2,3,4,5,np.nan,6,7,np.nan]
}
```

```python
df = pd.DataFrame(data)
scaler = MinMaxScaler()
```

```python
impu = SimpleImputer(strategy="mean")
```

```python
new_x = impu.fit_transform(df)
```

```python

scale = scaler.fit_transform(new_x)
```
