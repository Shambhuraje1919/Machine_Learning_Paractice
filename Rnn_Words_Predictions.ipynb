```python
def tokenize(text):
  text = text.lower()
  text = text.replace('?','')
  text = text.replace("'",'')
  return text.split()
```

```python
tokenize("Who wrote 'To Kill a Mockingbird'?"	)
```

```python
vocab = {'<UNK>':0}
```

```python
vocab
```

```python
def build_vocab(row):
  tockenized_que = tokenize(row['question'])
  tockenized_ans = tokenize(row['answer'])
  merged_tockens = tockenized_que+tockenized_ans
  for token in merged_tockens:
    if token not in vocab:
      vocab[token]  = len(vocab)

```

```python
df.apply(build_vocab,axis=1)
```

```python
def text_toNumrical(text ,vocab):
  indexed_text = []
  for tocken in tokenize(text):
    if tocken in vocab:
      indexed_text.append(vocab[tocken])
    else:
      indexed_text.append(vocab['<UNK>'])

  return indexed_text


```

```python
text_toNumrical("Shambhuraje is the what the boss",vocab)
```

```python
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import torch
```

```python
df.columns.unique()
```

```python
class QADataset(Dataset):
  def __init__(self ,df, vocab):
    self.df = df
    self.vocab = vocab


  def __len__(self):
    return self.df.shape[0]
  def __getitem__(self, index ):
    numerical_qustion = text_toNumrical(self.df.iloc[index]["question"],self.vocab)
    numerical_ans = text_toNumrical(self.df.iloc[index]["answer"],self.vocab)

    return torch.tensor(numerical_qustion) , torch.tensor(numerical_ans)
```

```python
dataset = QADataset(df,vocab)
```

```python
dataset
```

```python
vocab
```

```python
dataloader = DataLoader(dataset, shuffle= True, batch_size=1)
```

```python
for question, answer in dataloader:
  print(question, answer[0])
```

```python
class SimpleRnn(nn.Module):
  def __init__(self, vocab_size):
    super().__init__()
    self.embedding = nn.Embedding(vocab_size , embedding_dim=50)
    self.rnn = nn.RNN(50,64,batch_first=True)
    self.fc = nn.Linear(64,vocab_size)



  def forward(self, question):
    embedded_question = self.embedding(question)
    hidden, final = self.rnn(embedded_question)
    output = self.fc(final.squeeze(0))

    return output
```

```python
model = SimpleRnn(len(vocab))
```

```python
learning_rate = 0.001
epochs = 25
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(),learning_rate)
```

```python
# training loop
for epoch in range (epochs):
  total_loss =0
  for que,ans in dataloader:
    # zero grad
    optimizer.zero_grad()
    # forward pass
    op = model(que)
    # loss
    loss = criterion(op, ans[0])
    # backward
    loss.backward()
    # add new grad
    optimizer.step()

    total_loss = total_loss+loss.item()
  print(f"epoch{epoch+1},total Loss {total_loss}")


```
