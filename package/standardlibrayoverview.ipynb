FIle Operation -__- Read and Write Files

```python
## reada a whole file 
with open ('example.txt','r') as file:
    content = file.read()

    print(content)
```

```python
''' Read a File Line by Line'''
with open ('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
## Writing a  file 
with open ('example.txt', 'w') as file:
    file.write("Hello world\n")
    file.write("Hello Malak \n")
    
```

```python
with open ('example.txt', 'a') as file:
  file.write("Appending is taking place ")
  
```

```python
''' Writing a list of lines to a file'''
lines = ['first line \n', 'second line \n', 'thrird line malak \n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
    print()
```
