```python
import matplotlib.pyplot as plt
fruits = ['Apple', 'Mango', "banana", "orange","grapes"]
colors = ['lightblue','lightgreen','yellow','orange','purple']
explode = (0.1,0,0,0,0)
market_cap = [90,45,45,50,90]
plt.xlabel = fruits
plt.ylabel = market_cap
plt.title = 'Fruits Market Cap '
plt.bar(fruits,market_cap,color= 'purple')

plt.show()
plt.scatter (fruits,market_cap, color = 'yellow' , marker = "*",s=100)
plt.show()
plt.hist(market_cap, bins = 5, color = 'teal', edgecolor = 'violet')
plt.show()

plt.pie(market_cap,labels =fruits,autopct = "%1.1f%%", colors = colors, explode = explode)
```

SQl and SQLite in python
sql - is huge thing , lot of configuration ,
sqlite - no configuration , used in embedded database

```python
import sqlite3
```

```python
## Connect to sqllite data ba
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''CREATE TABLE IF NOT EXISTS employees (
    Id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
)''')
## commit changes
connection.commit()
```

```python
cursor.execute('''Select * from employees''')
```

```python
## Intsert the data in sqlite table
cursor.execute('''Insert into employees (name,age, department) values ('sam', 30, 'Malak')''')
cursor.execute('''Insert into employees (name,age,department) values('shambhu' ,21,'CEO')''')
```

```python
## Quraring the data from the table 
cursor.execute('Select * from employees')
rows = cursor.fetchall()
#Print the quearing data
for row in rows:
    print(row)
```

```python
## update
cursor.execute('''
               UPDATE employees
               Set age = 34
               where  name = "shambhu"
                ''')
connection.commit()
```

```python
## Deleting data from table
cursor.execute(''' Delete from employees where name = "shambhu"''')
connection.commit()
```
