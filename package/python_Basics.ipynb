```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [12,13,14,15,16]
plt.plot(x,y)
plt.xlabel("X-Axis")
plt.ylabel("Y-Axix")
plt.title("Simple line plot practice")
plt.show()
```

```python
plt.plot(x,y, color = "green" , linestyle = "--", linewidth = 3, marker = "o", markersize = 9)
plt.title("Customized Line Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.grid(True)
plt.show()
```

Multiple columsn and rows matplotlib

```python
x1 = [1,2,3,4,5]
y1 = [7,8,9,10,11]
y2 = [11,12,130,14,15]

plt.subplot(2,2,1)
plt.plot(x1,y1 , color = 'red' , marker = "o")
plt.title("Subplot 1")
#plt.show()
#plt.grid(True)

plt.subplot(2,2,2)
plt.plot(x1,y2 , color = "blue", marker = "o")
plt.title("Subplot 2")

plt.subplot(2,2,3)
plt.title("subplot 3")
plt.plot(x1,y2 , color = "goldenrod" , marker = "o")


plt.subplot(2,2,4)
plt.title("Subplot 4")
plt.plot(y1,y2 , color = "green" , marker = 'o')
plt.tight_layout()

```
