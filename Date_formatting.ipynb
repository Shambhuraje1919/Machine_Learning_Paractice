```python
import os
```

```python
import pandas as pd
import numpy as np
```

```python
path = "/orders (1).csv"
```

```python
df = pd.read_csv("/orders (1).csv")
```

```python
df.head()
```

```python
date = df
```

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date['year'] = date['date'].dt.year
```

```python
date.head()
```

```python
date['month'] = date['date'].dt.month_name()
df.head(2)
```

```python
date['date_day'] = date['date'].dt.day
date.head(2)
```

```python
date["No_Ofweek"] = date['date'].dt.day_of_week
date.head(2)
```

```python
date["day_name"] = date['date'].dt.day_name()
df.head()
```

```python
date = date.drop(["product_id","city_id"],axis=1)
```

```python
date.head(2)
```

```python
date['date_is_weekend'] = np.where(date['day_name'].isin(['Sunday', 'Saturday']), 1,0)
```
